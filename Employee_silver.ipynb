```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
df_emp_bronze = spark.read.table('emp_bronze')
```

```python
df_emp_silver = df_emp_bronze.withColumn('COMM',when(col('COMM').isNull() , lit(0)).otherwise(col('COMM')))
                                        
df_emp_silver.display()
```

## Write into table emp_silver

```python
df_emp_silver.write.saveAsTable('emp_silver')
```
